Números De +, - E *

```python
Nome = input('Informe O Seu Nome: ')
Idade = int(input('Informe A Sua Idade: '))
n1 = int(input('Primeiro Número: '))
n2 = int(input('Seguno Número: '))
Soma = n1 + n2
print('Olá {}, Idade {}, Soma {} + {} É Igual A {}'.format(Nome, Idade, n1, n2, Soma))
```

```python
Nome = input('Informe O Seu Nome: ')
Idade = int(input('Diga A Sua Idade: '))
n1 = int(input('Primeiro Número: '))
n2 = int(input('Segundo Número: '))
Menos = n1 - n2
print('Olá {}, Idade {},  {} - {} É Igual A {}'.format(Nome, Idade, n1, n2, Menos))
```
